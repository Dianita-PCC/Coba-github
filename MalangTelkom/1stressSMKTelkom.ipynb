# SLEEP_DURATION

```python
def train_test_split_np(X, y, test_size=0.2, seed=42):
    np.random.seed(seed) 
    idx = np.arange(len(X))
    np.random.shuffle(idx)
    
    n_test = int(len(X) * test_size)
    test_idx = idx[:n_test]
    train_idx = idx[n_test:]
    
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

X = df["Sleep_Duration"].to_numpy()
y = df["Stress_Level"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split_np(X, y)

print("Train:", len(X_train))
print("Test:", len(X_test))
```

```python
mean_x = np.mean(X_train)
mean_y = np.mean(y_train)

numerator = np.sum((X_train - mean_x) * (y_train - mean_y))
denominator = np.sum((X_train - mean_x) ** 2)

b1 = numerator / denominator
b0 = mean_y - b1 * mean_x

print("Slope (b1):", b1)
print("Intercept (b0):", b0)
```

```python
#rumus pred
y_pred = b0 + b1 * X_test
```

```python
#eval
def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))
    
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

def rmse(y_true, y_pred):
    return np.sqrt(mse(y_true,y_pred))

def r2(y_true,y_pred):
    ss_total = np.sum((y_true - np.mean(y_true))**2)
    ss_res = np.sum((y_true - y_pred)**2)
    return 1 - (ss_res/ss_total)

print("MAE:", mae(y_test,y_pred))
print("MSE:", mse(y_test,y_pred))
print("RMSE:", rmse(y_test,y_pred))
print("R2 Score:", r2(y_test,y_pred))
    
```

```python
sorted_idx = np.argsort(X_test)

X_sorted = X_test[sorted_idx]
y_pred_sorted = y_pred[sorted_idx]

plt.scatter(X_test, y_test, label="Actual Stress")
plt.plot(X_sorted, y_pred_sorted, label="Regression Line")

plt.xlabel("Sleep_Duration")
plt.ylabel("Stress_Level")
plt.title("Sleep_Duration vs Stress_Level (Testing Data)")
plt.legend()

plt.show()
```
