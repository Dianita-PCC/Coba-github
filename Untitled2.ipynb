```python
def train_test_split_np(X, y, test_size=0.2, seed=42): #membuat fungsi, data dibagi 20% untuk testing
    np.random.seed(seed) #mengunci untuk pembagian data test dan trainingy agar tetap sama
    idx = np.arange(len(X)) #urutan indeks data
    np.random.shuffle(idx)

    n_test = int(len(X) * test_size)
    test_idx = idx[:n_test]
    train_idx = idx[n_test:] #membgi index test dan train

    return X[train_idx], X[test_idx], y[train_idx], y[test_idx] #mengembalikan data train dan test

X = df["TV"].to_numpy()
y = df["Sales"].to_numpy() #Diubah menjadi numpy array supaya mudah dihitung.

X_train, X_test, y_train, y_test = train_test_split_np(X, y)

print("Train:", len(X_train)) #menampilkan jumlah data
print("Test:", len(X_test))
```
