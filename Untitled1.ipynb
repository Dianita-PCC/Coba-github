```python
# LINEAR REGRESSION FROM SCRATCH
# ============================================

class LinearRegressionScratch:
    """
    Linear Regression implementation from scratch using NumPy
    """
    def __init__(self):
        self.b0 = None  # intercept
        self.b1 = None  # slope
        self.X_mean = None
        self.y_mean = None
        
    def fit(self, X, y):
        """
        Train the linear regression model using least squares method
        
        Parameters:
        X: array-like, feature values (TV budget)
        y: array-like, target values (Sales)
        """

 # Convert to numpy arrays if needed
        X = np.array(X)
        y = np.array(y)
        
        # Calculate means
        self.X_mean = np.mean(X)
        self.y_mean = np.mean(y)

 # Calculate slope (b1) using formula: b1 = Σ((x - x̄)(y - ȳ)) / Σ((x - x̄)²)
        numerator = np.sum((X - self.X_mean) * (y - self.y_mean))
        denominator = np.sum((X - self.X_mean) ** 2)
        
        self.b1 = numerator / denominator

        # Calculate intercept (b0) using formula: b0 = ȳ - b1 * x̄
        self.b0 = self.y_mean - self.b1 * self.X_mean
        
        return self
    
    def predict(self, X):
        """
        Make predictions using the trained model
        
        Parameters:
        X: array-like, feature values to predict
        
        Returns:
        array: predicted values
        """
        X = np.array(X)
        return self.b0 + self.b1 * X
    
    def get_params(self):
        """Return model parameters"""
        return {'intercept': self.b0, 'slope': self.b1}
    
    def summary(self):
        """Print model summary"""
        print("\n" + "="*50)
        print("MODEL SUMMARY")
        print("="*50)
        print(f"Intercept (b0): {self.b0:.4f}")
        print(f"Slope (b1): {self.b1:.4f}")
        print(f"Equation: Sales = {self.b0:.4f} + {self.b1:.4f} * TV")
        
```

```python
 # TRAIN-TEST SPLIT FUNCTION
# ============================================

def train_test_split_manual(X, y, test_size=0.2, random_seed=42):
    """
    Manual train-test split without sklearn
    
    Parameters:
    X: array-like, features
    y: array-like, target
    test_size: float, proportion of test data
    random_seed: int, for reproducibility
    
    Returns:
    X_train, X_test, y_train, y_test
    """
    # Convert to numpy arrays
    X = np.array(X)
    y = np.array(y)
    
    # Set random seed for reproducibility
    np.random.seed(random_seed)
    
    # Create indices and shuffle
    n_samples = len(X)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    
    # Calculate split point
    split_point = int(n_samples * (1 - test_size))
    
    # Split indices
    train_indices = indices[:split_point]
    test_indices = indices[split_point:]
    
    # Split data
    X_train = X[train_indices]
    X_test = X[test_indices]
    y_train = y[train_indices]
    y_test = y[test_indices]
    
    return X_train, X_test, y_train, y_test
```

```python
# EVALUATION METRICS
# ============================================

def mean_absolute_error(y_true, y_pred):
    """Calculate Mean Absolute Error"""
    return np.mean(np.abs(y_true - y_pred))

def mean_squared_error(y_true, y_pred):
    """Calculate Mean Squared Error"""
    return np.mean((y_true - y_pred) ** 2)

def root_mean_squared_error(y_true, y_pred):
    """Calculate Root Mean Squared Error"""
    return np.sqrt(mean_squared_error(y_true, y_pred))

def r2_score(y_true, y_pred):
    """Calculate R-squared score"""
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - (ss_res / ss_tot)

def adjusted_r2_score(y_true, y_pred, n_features):
    """Calculate Adjusted R-squared"""
    n = len(y_true)
    r2 = r2_score(y_true, y_pred)
    return 1 - ((1 - r2) * (n - 1) / (n - n_features - 1))

```

```python
# IMPLEMENTATION
# ============================================

# Prepare data
X = df['TV'].values
y = df['Sales'].values

# Split data
X_train, X_test, y_train, y_test = train_test_split_manual(
    X, y, test_size=0.2, random_seed=42
)
```

```python
print("\n" + "="*50)
print("DATA SPLIT INFO")
print("="*50)
print(f"Total samples: {len(X)}")
print(f"Training samples: {len(X_train)} ({len(X_train)/len(X)*100:.1f}%)")
print(f"Testing samples: {len(X_test)} ({len(X_test)/len(X)*100:.1f}%)")
```

```python
# Create and train model
model = LinearRegressionScratch()
model.fit(X_train, y_train)

# Model summary
model.summary()
```

```python
# Make predictions
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)
```

```python
# Evaluate on training data
print("\n" + "="*50)
print("TRAINING SET PERFORMANCE")
print("="*50)
print(f"MAE: {mean_absolute_error(y_train, y_pred_train):.4f}")
print(f"MSE: {mean_squared_error(y_train, y_pred_train):.4f}")
print(f"RMSE: {root_mean_squared_error(y_train, y_pred_train):.4f}")
print(f"R²: {r2_score(y_train, y_pred_train):.4f}")
```

```python
# Evaluate on testing data
print("\n" + "="*50)
print("TESTING SET PERFORMANCE")
print("="*50)
print(f"MAE: {mean_absolute_error(y_test, y_pred_test):.4f}")
print(f"MSE: {mean_squared_error(y_test, y_pred_test):.4f}")
print(f"RMSE: {root_mean_squared_error(y_test, y_pred_test):.4f}")
print(f"R²: {r2_score(y_test, y_pred_test):.4f}")
print(f"Adjusted R²: {adjusted_r2_score(y_test, y_pred_test, 1):.4f}")
```
