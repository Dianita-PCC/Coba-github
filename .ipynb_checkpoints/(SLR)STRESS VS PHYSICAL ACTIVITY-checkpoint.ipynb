```python
X = df_new['Physical_Activity']
y = df_new['Stress_Level']
```

```python
xmx = (X - X.min())/(X.max() - X.min())
```

```python
split = int(0.8 * len(xmx))

x_train = xmx.iloc[:split]
x_test = xmx.iloc[split:]
y_train = y.iloc[:split]
y_test = y.iloc[split:]

x_trb = np.c_[np.ones(len(x_train)), x_train]
x_tsb = np.c_[np.ones(len(x_test)), x_test]

theta = np.linalg.inv(x_trb.T @ x_trb) @ x_trb.T @ y_train 
```

```python
ypred_tr = x_trb @ theta
ypred_ts = x_tsb @ theta

mse = np.mean(np.abs(y_test - ypred_ts))
print("MSE :", mse)
```

```python
ymean = np.mean(y_test)
ssres = np.sum((y_test - ypred_ts)** 2)
sstot = np.sum((y_test - ymean)** 2)
r2 = 1 - (ssres / sstot)
print("R-Square :", r2)
```

```python
print(x_train.head())
print(x_test.head())
print("")
print(y_train.head())
print(y_test.head())
```

```python
intercept = float(theta[0])
slope = float(theta[1])
```

```python
xmin = X.min()
xmax = X.max()
```

```python
def predict_stress() :
    x = float(entry.get())
    norm = (x - xmin) / (xmax - xmin)
    y = intercept + slope * norm
    acc = r2 * 100
    result_label.config(text=f"Predict : {y:.2f} | Acc : {acc:.2f}%")
                        
root = tk.Tk()
root.title("Stress Predict")

tk.Label(root, text="Physical Activity").pack()

entry = tk.Entry(root)
entry.pack()

tk.Button(root, text="Predict", command=predict_stress).pack()

result_label = tk.Label(root, text="")
result_label.pack()

root.mainloop()
```
