```python
X = df['YearsExperience']
y = df['Salary']
```

```python
x_bar = X.mean()
y_bar = y.mean()
```

```python
theta_1 = np.sum((X - x_bar) * (y - y_bar)) / np.sum((X - x_bar) ** 2)
theta_0 = y_bar - theta_1 * x_bar

print("Theta 1 (slope):", theta_1)
print("Theta 0 (intercept):", theta_0)
```

```python
y_pred = theta_0 + theta_1 * X
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X, y, label="Data Asli")
plt.plot(X, y_pred, label="Garis Regresi")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
mse = np.mean((y - y_pred)) ** 2
print("MSE :", mse)
```

```python
ss_total = np.sum((y - y_bar) ** 2)
ss_residual = np.sum((y - y_pred) ** 2)
```

```python
r2 = 1 - (ss_residual / ss_total)
print("R-squared:", r2)
```

```python
tahun_pengalaman = 5
prediksi_gaji = theta_0 + theta_1 * tahun_pengalaman
print("Prediksi gaji untuk 5 tahun pengalaman:", prediksi_gaji)
```
